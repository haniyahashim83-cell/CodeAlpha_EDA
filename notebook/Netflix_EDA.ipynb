## Visualization 1: Movies vs TV Shows

```python
import matplotlib.pyplot as plt

# Count Movies and TV Shows
type_count = df["type"].value_counts()

# Create figure
plt.figure(figsize=(6,4))

# Plot bar chart
type_count.plot(kind="bar", color=["red", "blue"])

# Add title and labels
plt.title("Movies vs TV Shows on Netflix")
plt.xlabel("Content Type")
plt.ylabel("Number of Titles")

# Save chart
plt.savefig("../images/movies_vs_tvshows.png", dpi=300)

# Show chart
plt.show()
```

## Visualization 2: Top 10 Countries

```python
import matplotlib.pyplot as plt

# Top 10 countries
top_countries = df["country"].dropna().str.split(", ").explode().value_counts().head(10)

plt.figure(figsize=(10,6))
top_countries.plot(kind="bar", color="green")

plt.title("Top 10 Countries on Netflix")
plt.xlabel("Country")
plt.ylabel("Number of Titles")

plt.xticks(rotation=45)

plt.savefig("../images/top_countries.png", dpi=300)

plt.show()
```

## Visualization 3: Content Released by Year

```python
release = df["release_year"].value_counts().sort_index()

plt.figure(figsize=(12,6))
release.plot()

plt.title("Netflix Content Released by Year")
plt.xlabel("Release Year")
plt.ylabel("Number of Titles")

plt.savefig("../images/release_year.png", dpi=300)

plt.show()
```

## Visualization 4: Content Ratings

```python
ratings = df["rating"].value_counts().head(10)

plt.figure(figsize=(10,6))
ratings.plot(kind="bar", color="orange")

plt.title("Most Common Ratings")
plt.xlabel("Rating")
plt.ylabel("Count")

plt.savefig("../images/ratings.png", dpi=300)

plt.show()
```

## Visualization 5: Top Genres

```python
genres = (
    df["listed_in"]
    .str.split(", ")
    .explode()
    .value_counts()
    .head(10)
)

plt.figure(figsize=(12,6))
genres.plot(kind="bar", color="purple")

plt.title("Top 10 Netflix Genres")
plt.xlabel("Genre")
plt.ylabel("Number of Titles")

plt.xticks(rotation=45)

plt.savefig("../images/genres.png", dpi=300)

plt.show()
```

## Visualization 6: Movies vs TV Shows Percentage

```python
type_count = df["type"].value_counts()

plt.figure(figsize=(6,6))

plt.pie(
    type_count,
    labels=type_count.index,
    autopct="%1.1f%%",
    startangle=90
)

plt.title("Movies vs TV Shows")

plt.savefig("../images/pie_chart.png", dpi=300)

plt.show()
```
